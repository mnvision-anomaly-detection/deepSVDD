```python
from PIL import Image
import numpy as np
import os
import matplotlib.pyplot as plt
import io
from rembg import remove
```

```python
#이미지 불러오기 + 전처리
input_folder = "/shared/cropped_dataset/normal"
```

```python
processed_images = []
```

```python
for filename in os.listdir(input_folder):
    if filename.endswith(('.png', '.jpg', '.jpeg')):  # 이미지 확장자 확인
        input_path = os.path.join(input_folder, filename)
        
        # 원본 이미지 파일 읽기
        with open(input_path, 'rb') as input_file:
            input_image = input_file.read()

        # 배경 제거
        output_image = remove(input_image)

        # PIL로 변환하여 numpy 배열로 변환
        img = Image.open(io.BytesIO(output_image))
        img_array = np.array(img)

        # 이미지 배열 리스트에 추가
        processed_images.append(img_array)

        print(f"Processed {filename}")
```

```python
if processed_images:
    first_image = processed_images[0]
    img_to_save = Image.fromarray(first_image)
    output_path = "/shared/DeepSVDD/output_image.png"
    img_to_save.save(output_path)
    print(f"First image saved as {output_path}")
```

```python
output_folder = "/shared/DeepSVDD/Alpha_images"
```

```python
for idx, img_array in enumerate(processed_images):
    img_to_save = Image.fromarray(img_array)
    output_path = os.path.join(output_folder, f"image_{idx + 1}.png")
    img_to_save.save(output_path)
    print(f"Saved image_{idx + 1}.png")
```
